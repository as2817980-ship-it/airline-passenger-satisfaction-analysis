```python

# Compare satisfaction by service rating and passenger class

for col in ["in_flight_wifi_service", "online_boarding"]:
    print(f"\n{'=' * 55}")
    print(f"Service: {col}")
    print("=" * 55)

    result = (
        df_clean.groupby(["class", col])["satisfaction"]
        .apply(lambda x: (x == "Satisfied").mean() * 100)
        .round(2)
        .rename("satisfaction_rate_pct")
        .reset_index()
    )

    print(result.to_string(index=False))
```

```python

boarding_results = (
    df_clean.groupby(["class", "online_boarding"])
    .agg(
        total_passengers=("satisfaction", "size"),
        satisfied_passengers=(
            "satisfaction",
            lambda x: (x == "Satisfied").sum()
        )
    )
    .reset_index()
)

boarding_results["satisfaction_rate_pct"] = (
    boarding_results["satisfied_passengers"]
    / boarding_results["total_passengers"] * 100
).round(2)

display(boarding_results.to_string(index=False))
```

```python

# Compare low ratings (1–3) for WiFi and online boarding by class.
# Rating 0 is excluded until its meaning is confirmed.

services = {
    "WiFi": "in_flight_wifi_service",
    "Online Boarding": "online_boarding",
}

rows = []

for service_name, col in services.items():
    temp = df_clean[df_clean[col].between(1, 3)]

    summary = (
        temp.groupby("class")
        .agg(
            passengers=("satisfaction", "size"),
            satisfied=("satisfaction",
                       lambda x: (x == "Satisfied").sum())
        )
        .reset_index()
    )

    summary["service"] = service_name
    summary["dissatisfied_pct"] = (
        100 * (summary["passengers"] - summary["satisfied"])
        / summary["passengers"]
    ).round(2)

    rows.append(summary)

comparison = pd.concat(rows, ignore_index=True)

display(
    comparison[
        ["service", "class", "passengers", "dissatisfied_pct"]
    ].sort_values(["class", "service"]).to_string(index=False)
)
```

```python

# Compare satisfaction by departure-delay band

df_delay = df_clean.copy()

df_delay["departure_delay_band"] = pd.cut(
    df_delay["departure_delay"],
    bins=[-float("inf"), 0, 15, 30, 60, float("inf")],
    labels=[
        "No delay",
        "1–15 minutes",
        "16–30 minutes",
        "31–60 minutes",
        "Over 60 minutes"
    ]
)

delay_summary = (
    df_delay.groupby("departure_delay_band", observed=False)
    .agg(
        passengers=("satisfaction", "size"),
        satisfied=("satisfaction",
                   lambda x: (x == "Satisfied").sum())
    )
    .reset_index()
)

delay_summary["satisfaction_rate_pct"] = (
    100 * delay_summary["satisfied"]
    / delay_summary["passengers"]
).round(2)

display(delay_summary.to_string(index=False))
```

```python

df_arrival = df_clean.copy()

df_arrival["arrival_delay_band"] = pd.cut(
    df_arrival["arrival_delay"],
    bins=[-float("inf"), 0, 15, 30, 60, float("inf")],
    labels=[
        "No delay",
        "1–15 minutes",
        "16–30 minutes",
        "31–60 minutes",
        "Over 60 minutes"
    ]
)

arrival_summary = (
    df_arrival.groupby("arrival_delay_band", observed=False)
    .agg(
        passengers=("satisfaction", "count"),
        satisfied=("satisfaction",
                   lambda x: (x == "Satisfied").sum())
    )
    .reset_index()
)

arrival_summary["satisfaction_rate_pct"] = (
    100 * arrival_summary["satisfied"]
    / arrival_summary["passengers"]
).round(2)

display(arrival_summary.to_string(index=False))
```

```python

import pandas as pd

service_columns = [
    "departure_and_arrival_time_convenience",
    "ease_of_online_booking",
    "check_in_service",
    "online_boarding",
    "gate_location",
    "on_board_service",
    "seat_comfort",
    "leg_room_service",
    "cleanliness",
    "food_and_drink",
    "in_flight_service",
    "in_flight_wifi_service",
    "in_flight_entertainment",
    "baggage_handling"
]

df_driver = df_clean.copy()

df_driver["satisfied_flag"] = (
    df_driver["satisfaction"] == "Satisfied"
).astype(int)

driver_results = []

for col in service_columns:
    valid = df_driver[[col, "satisfied_flag"]].dropna()

    driver_results.append({
        "service": col,
        "spearman_correlation": valid[col].corr(
            valid["satisfied_flag"], method="spearman"
        ),
        "average_rating": valid[col].mean(),
        "passengers": len(valid)
    })

driver_summary = pd.DataFrame(driver_results)

driver_summary["spearman_correlation"] = (
    driver_summary["spearman_correlation"].round(3)
)
driver_summary["average_rating"] = (
    driver_summary["average_rating"].round(2)
)

display(
    driver_summary.sort_values(
        "spearman_correlation", ascending=False
    ).to_string(index=False)
)
```

```python

# Combine satisfaction association and service performance

priority = driver_summary.copy()

priority["service"] = priority["service"].str.replace(
    "_", " ", regex=False
).str.title()

priority["importance_rank"] = priority[
    "spearman_correlation"
].rank(ascending=False, method="min").astype(int)

priority["performance_rank"] = priority[
    "average_rating"
].rank(ascending=True, method="min").astype(int)

display(
    priority.sort_values(
        ["importance_rank", "performance_rank"]
    )[
        [
            "service",
            "spearman_correlation",
            "average_rating",
            "importance_rank",
            "performance_rank"
        ]
    ].to_string(index=False)
)
```

```python

# Estimate how widespread low service ratings are.
# Ratings 1–3 are used; rating 0 is excluded pending validation.

service_columns = [
    "online_boarding",
    "in_flight_entertainment",
    "seat_comfort",
    "on_board_service",
    "leg_room_service",
    "cleanliness",
    "in_flight_wifi_service",
    "baggage_handling",
    "in_flight_service",
    "check_in_service",
    "food_and_drink",
    "ease_of_online_booking",
    "gate_location",
    "departure_and_arrival_time_convenience"
]

reach_rows = []

for col in service_columns:
    ratings = df_clean[col]
    valid = ratings.notna()
    low = ratings.between(1, 3)

    reach_rows.append({
        "service": col.replace("_", " ").title(),
        "valid_passengers": int(valid.sum()),
        "low_rating_passengers": int(low.sum()),
        "low_rating_pct": round(
            100 * low.sum() / valid.sum(), 2
        )
    })

reach_summary = pd.DataFrame(reach_rows)

display(
    reach_summary.sort_values(
        "low_rating_passengers", ascending=False
    ).to_string(index=False)
)
```

```python

# Recalculate service correlations excluding "not applicable" ratings (0)

service_columns = [
    "departure_and_arrival_time_convenience",
    "ease_of_online_booking",
    "check_in_service",
    "online_boarding",
    "gate_location",
    "on_board_service",
    "seat_comfort",
    "leg_room_service",
    "cleanliness",
    "food_and_drink",
    "in_flight_service",
    "in_flight_wifi_service",
    "in_flight_entertainment",
    "baggage_handling"
]

df_driver = df_clean.copy()

df_driver["satisfied_flag"] = (
    df_driver["satisfaction"] == "Satisfied"
).astype(int)

driver_results = []

for col in service_columns:
    valid = df_driver.loc[
        df_driver[col].between(1, 5),
        [col, "satisfied_flag"]
    ]

    driver_results.append({
        "service": col,
        "spearman_correlation": valid[col].corr(
            valid["satisfied_flag"], method="spearman"
        ),
        "average_rating": valid[col].mean(),
        "passengers_rated": len(valid),
        "not_applicable": int((df_driver[col] == 0).sum())
    })

driver_summary = pd.DataFrame(driver_results)

driver_summary["spearman_correlation"] = (
    driver_summary["spearman_correlation"].round(3)
)
driver_summary["average_rating"] = (
    driver_summary["average_rating"].round(2)
)

display(
    driver_summary.sort_values(
        "spearman_correlation", ascending=False
    ).to_string(index=False)
)
```

```python

df_segments = df_clean.copy()

df_segments["age_band"] = pd.cut(
    df_segments["age"],
    bins=[0, 25, 40, 60, float("inf")],
    labels=["18–25", "26–40", "41–60", "60+"],
    include_lowest=True
)

segment_columns = [
    "class",
    "type_of_travel",
    "customer_type",
    "age_band"
]

for col in segment_columns:
    print(f"\nSatisfaction by {col}")

    result = (
        df_segments.groupby(col, observed=False)
        .agg(
            passengers=("satisfaction", "size"),
            satisfied=("satisfaction",
                       lambda x: (x == "Satisfied").sum())
        )
        .reset_index()
    )

    result["satisfaction_rate_pct"] = (
        100 * result["satisfied"] / result["passengers"]
    ).round(2)

    display(result.to_string(index=False))
```

```python

service_columns = [
    "online_boarding",
    "in_flight_entertainment",
    "in_flight_wifi_service",
    "seat_comfort",
    "on_board_service",
    "leg_room_service",
    "cleanliness",
    "baggage_handling",
    "in_flight_service",
    "check_in_service",
    "food_and_drink",
    "ease_of_online_booking"
]

segment_driver_rows = []

for passenger_class in df_clean["class"].dropna().unique():
    class_df = df_clean[df_clean["class"] == passenger_class]

    for col in service_columns:
        valid = class_df.loc[
            class_df[col].between(1, 5),
            [col, "satisfaction"]
        ]

        satisfaction_flag = (
            valid["satisfaction"] == "Satisfied"
        ).astype(int)

        segment_driver_rows.append({
            "class": passenger_class,
            "service": col,
            "correlation": valid[col].corr(
                satisfaction_flag, method="spearman"
            ),
            "average_rating": valid[col].mean(),
            "passengers_rated": len(valid)
        })

segment_drivers = pd.DataFrame(segment_driver_rows)

segment_drivers["correlation"] = (
    segment_drivers["correlation"].round(3)
)
segment_drivers["average_rating"] = (
    segment_drivers["average_rating"].round(2)
)

display(
    segment_drivers.sort_values(
        ["class", "correlation"], ascending=[True, False]
    ).to_string(index=False)
)
```

```python

age_results = (
    df_segments.groupby("age_band", observed=False)
    .agg(
        passengers=("satisfaction", "size"),
        satisfied=("satisfaction",
                   lambda x: (x == "Satisfied").sum())
    )
    .reset_index()
)

age_results["satisfaction_rate_pct"] = (
    100 * age_results["satisfied"] / age_results["passengers"]
).round(2)

display(age_results.to_string(index=False))
```

```python

# Build a preliminary candidate ranking from existing results

candidates = driver_summary[
    ["service", "spearman_correlation",
     "average_rating", "passengers_rated"]
].copy()

candidates["service"] = candidates["service"].str.replace(
    "_", " ", regex=False
).str.title()

candidates["low_rating_passengers"] = candidates["service"].map(
    reach_summary.set_index(
        reach_summary["service"].str.replace(
            "_", " ", regex=False
        ).str.title()
    )["low_rating_passengers"]
)

display(
    candidates.sort_values(
        "spearman_correlation", ascending=False
    ).to_string(index=False)
)
```

```python

# Build a transparent screening score for service priorities.
# This is NOT a causal impact or ROI estimate.

priority = driver_summary.copy()

priority["service"] = priority["service"].str.replace(
    "_", " ", regex=False
).str.title()

reach = reach_summary.copy()
reach["service"] = reach["service"].str.replace(
    "_", " ", regex=False
).str.title()

priority = priority.merge(
    reach[["service", "low_rating_pct"]],
    on="service",
    how="left"
)

# Normalize importance to 0–1 across services
priority["importance_score"] = (
    priority["spearman_correlation"]
    - priority["spearman_correlation"].min()
) / (
    priority["spearman_correlation"].max()
    - priority["spearman_correlation"].min()
)

# Normalize performance gap and low-rating reach to 0–1
priority["performance_gap_score"] = (
    5 - priority["average_rating"]
) / 4

priority["reach_score"] = priority["low_rating_pct"] / 100

# Equal weights for an initial transparent screening score
priority["screening_score"] = 100 * (
    priority["importance_score"] / 3
    + priority["performance_gap_score"] / 3
    + priority["reach_score"] / 3
)

display(
    priority.sort_values(
        "screening_score", ascending=False
    )[
        [
            "service",
            "spearman_correlation",
            "average_rating",
            "low_rating_pct",
            "screening_score"
        ]
    ].round(2).to_string(index=False)
)
```
